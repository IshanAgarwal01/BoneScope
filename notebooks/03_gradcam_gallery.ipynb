*Educational project, not for clinical use.* For each class: 2 correct and 2 wrong test cells
of one fold, with the heat map for the predicted class. **Border heat** = share of heat in the
outer 15% of the crop; > 0.6 suggests the model looked at background or a neighbouring cell
(possible shortcut). Record those cases for the README's limitations.
